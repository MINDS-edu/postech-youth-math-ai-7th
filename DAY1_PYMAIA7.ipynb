❓누가 가장 좋은 값을 찾았는지 토의해 봅시다.

아마 여러분들은 아래와같이 예측값과 실재값의 차이를 줄여야한다는 직관에 따라 파라미터를 조절하고 있었을 것입니다.

```python
w = 1.9273372267001303# @param {type:"number"}
b = -69.44242000096528# @param {type:"number"}
m,c = w,b
fcn_estimate = lambda x: m*x + c
xs = df['height'].values
ys = df['shoe size'].values
def sort_by_x(x_unsorted, y_unsorted):
    """
    Sorts two lists/arrays based on the values in the first list/array.

    Args:
        x_unsorted: The list or array to sort by.
        y_unsorted: The list or array to sort according to the order of x_unsorted.

    Returns:
        A tuple containing the sorted x_unsorted and y_unsorted.
    """
    # Combine x and y into pairs, sort by x, then separate
    sorted_pairs = sorted(zip(x_unsorted, y_unsorted))
    x_sorted, y_sorted = zip(*sorted_pairs)
    return np.array(x_sorted), np.array(y_sorted)

# Example usage (optional - can be removed or commented out)
# x_unsorted = [5, 2, 8, 1, 9]
# y_unsorted = [10, 4, 16, 2, 18]
# x_sorted, y_sorted = sort_by_x(x_unsorted, y_unsorted)
# print("Sorted x:", x_sorted)
# print("Sorted y:", y_sorted)
def visualize_MSE(fcn_estimate,xs,ys,l_compare=100):
  l = l_compare
  xs,ys = sort_by_x(xs,ys)
  fcn = fcn_estimate
  fig, axs = plt.subplots(1, 2, layout='constrained',figsize=(10, 5))
  plt.sca(axs[0])
  plt.plot(xs,ys,'o',label='Actual Data')
  x_smooth = np.linspace(xs.min(), xs.max(), 100)
  plt.plot(x_smooth,fcn(x_smooth),label='Predict')
  plt.ylabel('Shoe Size')
  plt.xlabel('Height')
  xs = np.array(xs)
  ys = np.array(ys)
  errors = ys - fcn(xs)
  clr_error = []
  for i, error in enumerate(errors):
      if error >= 0:
          clr = 'red'
      else:
          clr = 'blue'
      plt.plot([xs[i], xs[i]], [fcn(xs[i]), ys[i]], color=clr, linestyle='-')
      clr_error.append(clr)
  plt.legend()
  plt.title('Visualize error')





  plt.sca(axs[1])
  # 정사각형 넓이 리스트
  square_areas = errors*errors

  # 정사각형 색상 리스트
  colors = ['red', 'blue', 'green', 'yellow', 'purple']

  # 넓이 합 계산
  total_area = sum(square_areas)

  # 시각화 시작
  ax = axs[1]

  # 정사각형 그리기
  start_x = 0

  for i, area in enumerate(square_areas):
    #area = abs(errors[i])
      side = np.sqrt(area)  # 정사각형 한 변의 길이 계산
      rect = plt.Rectangle((start_x, 0), side, side, facecolor=clr_error[i])
      ax.add_patch(rect)
      start_x += side  # 다음 정사각형 시작 위치 설정

  # 텍스트 추가 (넓이 합 표시)
  #plt.text(0.5, -0.1, f'Total Area: {total_area}',
  #         horizontalalignment='center', verticalalignment='center',
  #         transform=ax.transAxes, fontsize=12)

  # 그래프 설정


  #ax.set_xlim([0, 100])
  #ax.set_ylim([0, 100])  # y축 범위 설정
  ax.set_aspect('equal')  # 정사각형 모양 유지
  plt.title('Sum of Areas of Squares')
  # plt.xlabel('Width')

  plt.plot([0, l, l, 0, 0],[0, 0, l, l, 0],'--',label=f'RSS={l}$^2$')
  lsum = np.sqrt(square_areas.sum())
  xs_sum = np.array([0,1,1,0,0])
  ys_sum = np.array([0,0,1,1,0])
  plt.plot(xs_sum*lsum,ys_sum*lsum,label=f"RSS={lsum:.2f}$^2$")
  plt.legend()
  plt.xlabel(f'MSE=RSS/n={lsum**2/len(xs):.4f}')

visualize_MSE(fcn_estimate,xs,ys)
```

### 목적함수 도입

$$\textrm{MSE} = \frac{1}{n}\underbrace{\sum_i(y_i^{예측}-y_i^{관측})^2}_{\text{RSS}}$$

### 시그모이드 함수 (Sigmoid Function)

**시그모이드 함수**는 S자 형태를 띠는 비선형 함수로, 어떤 입력 값이 주어지더라도 그 결과를 0과 1 사이의 값으로 압축(변환)해 줍니다. 이러한 특성 때문에 특히 이진 분류 문제에서 특정 클래스에 속할 **확률**을 모델링하는 데 널리 사용됩니다.

**수학적 표현:**
시그모이드 함수는 다음과 같은 수식으로 정의됩니다.

$$ \sigma(x) = \frac{1}{1 + e^{-x}} $$

여기서:
*   $e$는 자연상수(약 2.718)입니다.
*   $x$는 어떤 실수 값도 될 수 있는 입력 값입니다 (예: 선형 회귀 모델의 출력 값).

**주요 특징:**
1.  **출력 범위**: 함수의 출력 값은 항상 0과 1 사이에 있습니다. $x$가 매우 큰 양수일 때는 1에 가까워지고, $x$가 매우 큰 음수일 때는 0에 가까워집니다.
2.  **S자 형태**: 그래프를 그리면 부드러운 S자 곡선 형태를 보입니다.
3.  **확률 해석**: 로지스틱 회귀에서는 선형 모델의 결과($wx+b$)를 시그모이드 함수에 통과시켜 0과 1 사이의 값을 얻습니다. 이 값은 특정 이벤트(예: 성별이 '남'일 확률)가 발생할 확률로 해석됩니다.

**로지스틱 회귀에서의 역할:**
로지스틱 회귀는 선형 회귀의 예측 값을 시그모이드 함수에 넣어 확률로 변환한 후, 이 확률을 바탕으로 범주를 분류합니다. 예를 들어, 시그모이드 함수의 출력 값이 0.5 이상이면 '남성', 0.5 미만이면 '여성'으로 분류하는 방식입니다.

👉시그모이드 함수(w,b를 바꿔봅시다)

```python
w = 1# @param {type:"number"}
b = -170# @param {type:"number"}
m,c = w,b
fcn_estimate = lambda x: 1/(np.exp(-(m*x + c))+1)
xs = df['height'].values
ys = np.float64((df['sex'].values=='male'))
visualize_MSE(fcn_estimate,xs,ys)
```

사실 로지스틱 회귀는 MSE보단 크로스엔트로피라는 목적함수를 쓰는 게 일반적입니다.

### 선형 회귀 vs 로지스틱 회귀

**선형 회귀 (Linear Regression)**
*   **목적**: 연속적인 숫자 값을 예측하는 데 사용됩니다. 예를 들어, 집의 가격, 주식 가격, 사람의 키와 같이 '얼마나', '몇 개'와 같이 연속된 값을 예측할 때 적합합니다.
*   **출력**: 어떤 범위든 가질 수 있는 연속적인 값입니다.
*   **예시**: 키를 통해 발사이즈를 예측하는 것.

**로지스틱 회귀 (Logistic Regression)**
*   **목적**: 범주형 값을 예측하는 데 사용됩니다. 즉, 어떤 항목이 특정 범주에 속할 확률을 예측하고, 그 확률을 기반으로 범주를 분류합니다. 예를 들어, 이메일이 스팸인지 아닌지(2가지 범주), 고객이 제품을 구매할지 안 할지(2가지 범주) 등을 예측할 때 적합합니다.
*   **출력**: 0과 1 사이의 확률 값입니다. 이 확률은 특정 범주에 속할 가능성을 나타내며, 일반적으로 0.5를 기준으로 특정 범주로 분류합니다.
*   **예시**: 발사이즈를 통해 성별(남/여)을 예측하는 것.

### 발사이즈로 성별 예측 문제에 로지스틱 회귀가 더 적합한 이유

1.  **목표 변수의 성격**: 성별은 '남' 또는 '여'와 같은 이산적인(범주형) 값입니다. 선형 회귀는 이처럼 0과 1로 명확히 나뉘는 범주를 직접 예측하기 어렵습니다. 선형 회귀의 예측 값은 특정 범위에 한정되지 않는 연속적인 값이기 때문에, 이를 '남' 또는 '여'와 같은 이산적인 범주로 변환하는 데 문제가 발생할 수 있습니다.

2.  **확률적 접근**: 로지스틱 회귀는 선형 함수의 결과를 시그모이드(Sigmoid) 함수에 통과시켜 0과 1 사이의 확률 값으로 변환합니다. 이 확률 값을 특정 기준(예: 0.5)과 비교하여 '남' 또는 '여'와 같은 범주로 분류합니다. 이는 성별과 같이 이진 분류 문제에 대한 자연스럽고 직관적인 접근 방식입니다.

3.  **오차 함수 (Cost Function)**: 선형 회귀에서 사용되는 MSE(평균 제곱 오차)는 분류 문제에 적용할 경우 적절한 학습을 방해할 수 있습니다. 로지스틱 회귀에서는 주로 '크로스 엔트로피(Cross-Entropy)'와 같은 오차 함수를 사용하여 모델의 예측 확률이 실제 레이블에 얼마나 가까운지를 측정하며, 이는 분류 문제에 더 효과적입니다.
